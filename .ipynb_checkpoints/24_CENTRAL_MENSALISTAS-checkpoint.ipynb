```python
# Substituindo os valores nulos na coluna Contrato por 241294
df['Contrato'] = df['Contrato'].fillna(241294)

# Após preencher, garantimos que a coluna seja do tipo inteiro 
# para que o número não fique aparecendo como 241294.0
df['Contrato'] = df['Contrato'].astype(int)
```

```python
import pandas as pd


# 1. Datas (Convertendo com padrão brasileiro)
df['Dt.Vecto.'] = pd.to_datetime(df['Dt.Vecto.'], dayfirst=True, errors='coerce')
df['Dt.Pagto.'] = pd.to_datetime(df['Dt.Pagto.'], dayfirst=True, errors='coerce')

# 2. Números Inteiros (N. Baixa e Cd.Cobrança não têm nulos no print)
df['N. Baixa'] = df['N. Baixa'].astype(int)
df['Cd.Cobrança'] = df['Cd.Cobrança'].astype(int)

# 3. Contrato (Nullable Integer para lidar com o NaN que vimos na linha 3)
df['Contrato'] = pd.to_numeric(df['Contrato'], errors='coerce').astype('Int64')

# 4. Valores Financeiros (Garantindo que são floats antes de qualquer formatação)
df['R$ Boleto'] = pd.to_numeric(df['R$ Boleto'], errors='coerce').astype(float)
df['R$ Pgto'] = pd.to_numeric(df['R$ Pgto'], errors='coerce').astype(float)

# 5. Colunas de Texto
cols_texto = ['Unidade', 'Cliente', 'Status', 'Compet.']
df[cols_texto] = df[cols_texto].astype(str)

df
```
