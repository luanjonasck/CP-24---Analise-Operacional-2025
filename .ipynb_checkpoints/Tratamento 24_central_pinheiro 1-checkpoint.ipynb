```python
df = pd.read_csv('24_CENTRAL_PINHEIRO.csv - base.csv', sep=None, engine='python')
```

```python
df = df.dropna(how='all').reset_index(drop=True)
df
```

```python
def clean_currency(x):
    if pd.isna(x): return 0.0
    if isinstance(x, str):
        clean_val = x.replace('R$', '').replace(' ', '').replace('.', '').replace(',', '.')
        try: return float(clean_val)
        except: return 0.0
    return float(x)

for col in ['Subtotal', 'Valor Pago', 'Vl. Desconto', 'Vl. Selo', 'Vl. Convenio']:
    df[col] = df[col].apply(clean_currency)
```

```python
df
```

```python
df['Entrada'] = pd.to_datetime(df['Entrada'], errors='coerce')
```

```python
df['Saida'] = pd.to_datetime(df['Saida'], errors='coerce')
```

```python
df['Stay_Min'] = (df['Saida'] - df['Entrada']).dt.total_seconds() / 60
```

```python
df['Stay_Min'] = df['Stay_Min'].fillna(0)
```

```python
df['Stay_Hour'] = (df['Stay_Min'] / 60).apply(lambda x: int(x) + 1 if (x > 0 and x % 1 > 0) else int(x))
```

```python
cols_financeiras = ['Subtotal', 'Valor Pago', 'Vl. Desconto', 'Vl. Selo', 'Vl. Convenio']
for col in cols_financeiras:
    df[col] = df[col].apply(clean_currency)
```

```python
import pandas as pd

# 1. Renomear as colunas Unnamed
df = df.rename(columns={
    'Unnamed: 6': 'Contrato',
    'Unnamed: 7': 'Cliente'
})

# 2. Tratar e transformar as colunas financeiras em Decimal (float)
# Primeiro garantimos que o valor seja texto, limpamos e convertemos
colunas_financeiras = ['Subtotal', 'Valor Pago']

for col in colunas_financeiras:
    # Caso os dados venham com vírgula (padrão BR), trocamos por ponto para o Python entender
    df[col] = df[col].astype(str).str.replace(',', '.')
    
    # Transformamos em numérico (float)
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# Verificando se os tipos mudaram para float64
print(df[['Contrato', 'Cliente', 'Subtotal', 'Valor Pago']].dtypes)
```

```python
df.info()
```

```python
# 1. Converte as colunas para datetime
# O format='mixed' ajuda se houver formatos diferentes na mesma coluna (comum em versões recentes do pandas)
df['Entrada'] = pd.to_datetime(df['Entrada'], dayfirst=False, errors='coerce')
df['Saida'] = pd.to_datetime(df['Saida'], dayfirst=False, errors='coerce')
```

```python
df
```

```python
df['Subtotal'] = df['Subtotal'].apply(lambda x: f"{x:.2f}".replace('.', ','))
df['Valor Pago'] = df['Valor Pago'].apply(lambda x: f"{x:.2f}".replace('.', ','))
```

```python
df
```

```python
# 3. Tratamento de Datas (Corrigindo o Warning de formato)
# dayfirst=False pois o padrão do seu arquivo parece ser MM/DD/YY (ex: 12/28/24)
df['Entrada_DT'] = pd.to_datetime(df['Entrada'], errors='coerce')
df['Saida_DT'] = pd.to_datetime(df['Saida'], errors='coerce')

# 4. Cálculo de Permanência com tratamento para NaN
# Calculamos os minutos
df['Permanencia_Minutos'] = (df['Saida_DT'] - df['Entrada_DT']).dt.total_seconds() / 60
df['Permanencia_Minutos'] = df['Permanencia_Minutos'].fillna(0)

# Cálculo de Horas arredondado para cima (Corrigindo o ValueError)
def calcular_hora_cheia(minutos):
    if minutos <= 0:
        return 0
    # math.ceil manual para evitar erros de tipo
    horas = minutos / 60
    if horas % 1 > 0:
        return int(horas) + 1
    return int(horas)

df['Permanencia_Horas'] = df['Permanencia_Minutos'].apply(calcular_hora_cheia)
```
